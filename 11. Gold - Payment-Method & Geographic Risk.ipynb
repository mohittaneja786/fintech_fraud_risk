# Payment-Method & Geographic Risk

```python
%sql
-- Q11: Fraud rate by payment method
SELECT
  f.payment_method,
  COUNT(DISTINCT f.transaction_id) AS txn_count,
  COUNT(DISTINCT fa.alert_id) AS fraud_count,
  ROUND(COUNT(DISTINCT fa.alert_id) * 100.0 / COUNT(DISTINCT f.transaction_id), 3) AS fraud_rate_pct
FROM fintech_fraud_risk.gold.fact_transactions f
LEFT JOIN fintech_fraud_risk.silver.fraud_alerts fa ON f.transaction_id = fa.transaction_id
GROUP BY f.payment_method
ORDER BY fraud_rate_pct DESC;
```

```python
%sql

-- Q12: Fraud rate by transaction location (geographic risk)
SELECT
  f.location,
  COUNT(DISTINCT f.transaction_id) AS txn_count,
  COUNT(DISTINCT fa.alert_id) AS fraud_count,
  ROUND(COUNT(DISTINCT fa.alert_id) * 100.0 / COUNT(DISTINCT f.transaction_id), 3) AS fraud_rate_pct
FROM fintech_fraud_risk.gold.fact_transactions f
LEFT JOIN fintech_fraud_risk.silver.fraud_alerts fa ON f.transaction_id = fa.transaction_id
GROUP BY f.location
ORDER BY fraud_rate_pct DESC;
```
