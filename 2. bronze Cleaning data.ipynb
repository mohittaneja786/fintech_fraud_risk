```python
base	= "/Volumes/fintech_fraud_risk/bronze/vol_bronze"

def	load_full(path_glob,	table_name):

    df	=	(
                spark.read.option("header",True)
                .option("inferSchema",	True)
                .csv(path_glob)
                .select("*","_metadata.file_name","_metadata.file_size")
            )
    
    df.write.mode("overwrite").format("delta")\
        .option("overwriteSchema", True)\
        .saveAsTable(f"fintech_fraud_risk.bronze.{table_name}")

    print(table_name,	df.count())

load_full(f"{base}/bronze_chargebacks.csv",	"bronze_chargebacks")
load_full(f"{base}/bronze_customers.csv",	"bronze_customers")
load_full(f"{base}/bronze_devices.csv","bronze_devices")
load_full(f"{base}/bronze_fraud_alerts.csv","bronze_fraud_alerts")
load_full(f"{base}/bronze_login_events.csv","bronze_login_events")
load_full(f"{base}/bronze_merchants.csv","bronze_merchants")
load_full(f"{base}/bronze_transactions_historical.csv","bronze_transactions")
```
